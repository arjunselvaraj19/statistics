# Statistics Task

```python
# Import Libs
from scipy import stats
import numpy as np
from statsmodels.stats.proportion import proportions_ztest
```

## 1. The historical average score in a certification exam is 65. A training institute introduces a new course. A random sample of 50 students trained under the new course has a mean score of 68 and a standard deviation of 10. At the 5% level of significance, test whether the new course improves performance.

```python

observation_value = np.random.normal(loc=68, scale=5.0, size=50).astype("int")
t_stats, p_value = stats.ttest_1samp(a = observation_value, popmean = 65, alternative="greater")
p_value = p_value / 2 #
print("P-Value: ", p_value.round(5))
if p_value < 0.05: # 5% Significance Level
    print("We reject H0 (Null Hypo.) and consider Alternative Hypothesis")
else:
    print("We do not reject H0 (Null Hypo.)")
```

##### **Result:**
**Test:** One Sample T-Test, Right Tailed (1-tailed)  
**H0:** = 65  
**H1:** > 65  
**Business Conclusion:** At the 5% significance level, there is a sufficient evidence that s are taking longer than claimed 30 mins. So the support team is not meeting its SLA.

## 3. A company claims that 60% of its users are satisfied with its product. In a survey of 200 users, 134 users report satisfaction. At the 5% level of significance, test whether customer satisfaction is higher than claimed.

```python

z_stats, p_value = proportions_ztest(count = 134, nobs = 200, value= 0.60, alternative='larger', prop_var=0.60)

p_value = p_value / 2 #
print("P-Value: ", p_value.round(5))
if p_value < 0.05: # 5% Significance Level
    print("We reject H0 (Null Hypo.) and consider Alternative Hypothesis")
else:
    print("We do not reject H0 (Null Hypo.)")
```

##### **Result:**
**Test:** One Sample Proportion Z-Test, Right Tailed (1-tailed)  
**H0:** = 0.60  
**H1:** > 0.60  
**Business Conclusion:** At the 5% significance level, there is a sufficient evidence that customer satisfaction (67%). The company can support its satisfaction claim.

## 5. Two teaching methods are used to train employees. A sample of 30 employees trained using Method A has a mean score of 75 with a standard deviation of 12. A sample of 28 employees trained using Method B has a mean score of 70 with a standard deviation of 10. At the 5% level of significance, test whether there is a difference in average scores between the two methods.

```python

# A sample
mean_a = 75
std_a  = 12
size_a = 30

# B sample
mean_b = 70
std_b  = 10
size_b = 28

significance_level = 0.05

t_statistic, p_value = stats.ttest_ind_from_stats(mean1=mean_a, std1=std_a, nobs1=size_a,
                                                  mean2=mean_b, std2=std_b, nobs2=size_b,
                                                  equal_var=True, alternative = "two-sided")

# Actuall we shd use equal_var is False bcoz we have diff in Std. dev but there is no much diff so we used True here

print("T-statistic: ", round(t_statistic, 4))
print("P-Value:     ", round(p_value, 5))

if p_value < significance_level:
    print("We reject H0 (Null Hypo.) and consider Alternative Hypothesis")
else:
    print("We do not reject H0 (Null Hypo.)")
```

##### **Result:**
**Test:** Two Sample T-Test, Two Tailed  
**H0:** uA = uB
**H1:** uA != uB  
**Business Conclusion:** At the 5% significance level, there is no significant difference between the two teaching methods. So either method can be used.

## 9. A company wants to compare the average productivity of employees trained using three different training programs. Program A has 25 employees with a mean productivity score of 80 and a standard deviation of 8. Program B has 30 employees with a mean of 75 and a standard deviation of 7. Program C has 28 employees with a mean of 82 and a standard deviation of 6. At the 5% level of significance, test whether there is a difference in mean productivity across the programs.

```python


program_means = np.array([80, 75, 82])
program_variances = np.array([8, 7, 6]) ** 2    # variance = std squared
program_sizes = np.array([25, 30, 28])
significance_level = 0.05
from statsmodels.stats.oneway import anova_generic

result = anova_generic(means = program_means, variances = program_variances, nobs = program_sizes, use_var='equal', welch_correction=False)

print("F-statistic: ", round(result.statistic, 4))
print("P-Value:     ", round(result.pvalue, 5))

if result.pvalue < significance_level:
    print("We reject H0 (Null Hypo.) and consider Alternative Hypothesis")
else:
    print("We do not reject H0")
```

##### **Result:**
**Test:** ANOVA One way  
**H0:** uA = uB = uC
**H1:** At least one mean is significantly disfferent  
**Business Conclusion:** At the 5% significance level, at least one program differs from the others. Based on the other Tukey's test, Program B mean is significantly lower than the other two.

```python
import numpy as np
from statsmodels.stats.multicomp import pairwise_tukeyhsd

np.random.seed(42)

program_a = np.random.normal(loc=80, scale=8, size=25).astype("int")
program_b = np.random.normal(loc=75, scale=7, size=30).astype("int")
program_c = np.random.normal(loc=82, scale=6, size=28).astype("int")

productivity_scores = np.concatenate([program_a, program_b, program_c])
productivity_scores

program_labels = ["A"] * 25 + ["B"] * 30 + ["C"] * 28

tukey_result = pairwise_tukeyhsd(endog=productivity_scores, groups=program_labels, alpha=0.05)
print(tukey_result)
```

## Conclusion: Program B is significantly different from both A and C (reject = True). Programs A and C do not differ significantly. This matches the manual Tukey result, so Program A or C is the better choice.

## 10. A company wants to test whether gender and product preference are independent. The survey results are as follows:
●	**Male:** 60 prefer Product A, 40 prefer Product B  
●	**Female:** 50 prefer Product A, 50 prefer Product B  
At the 5% level of significance, test for independence.

```python
import numpy as np
from scipy.stats import chi2_contingency

observed_counts = np.array([[60, 40],    # Male:   Product A, Product B
                            [50, 50]])   # Female: Product A, Product B
significance_level = 0.05

chi_square_statistic, p_value, degrees_of_freedom, expected_counts = chi2_contingency(observed_counts)

print("Chi-Square Statistic: ", round(chi_square_statistic, 4))
print("P-Value:              ", round(p_value, 5))
print("Degrees of Freedom:   ", degrees_of_freedom)

if p_value < significance_level:
    print("We reject H0 (Null Hypo.) and consider Alternative Hypothesis")
else:
    print("We do not reject H0")
```

##### **Result:**
**Test:** Chi Square Test of Independence  
**H0:** Gender and Product preference are independent  
**H1:** Gender and Product preference are not independent  
**Business Conclusion:** At the 5% significance level, there is no significant difference between the product and gender. Product preference does not depend on product. So same marketing strategy can be followed.

## 11. A  hospital studies whether smoking status is related to disease occurrence. The data collected is as follows:
●	**Smokers:** 70 have the disease, 30 do not  
●	**Non-smokers:** 40 have the disease, 60 do not  
At the 5% level of significance, test whether smoking status and disease occurrence are independent. 


```python
import numpy as np
from scipy.stats import chi2_contingency

observed_counts = np.array([[70, 30], 
                            [40, 60]])
significance_level = 0.05

chi_square_statistic, p_value, degrees_of_freedom, expected_counts = chi2_contingency(observed_counts)

print("Chi-Square Statistic: ", round(chi_square_statistic, 4))
print("P-Value:              ", round(p_value, 5))
print("Degrees of Freedom:   ", degrees_of_freedom)

if p_value < significance_level:
    print("We reject H0 (Null Hypo.) and consider Alternative Hypothesis")
else:
    print("We do not reject H0")
```

##### **Result:**
**Test:** Chi Square Test of Independence  
**H0:** Smoking habit and Disease are independent (no relation)  
**H1:** Smoking habit and Disease are not independent (related)    
**Business Conclusion:** At the 5% significance level, smoking habit and desease occurence are related. Disease is more common among smokers than non-smokers.

## 12.	A dice manufacturer claims that a die is fair. The die is rolled 120 times, and the observed frequencies are:
●	Face 1: 18  
●	Face 2: 22  
●	Face 3: 20  
●	Face 4: 25  
●	Face 5: 17  
●	Face 6: 18  
At the 5% level of significance, test whether the die is fair.



```python
from scipy.stats import chisquare

observed_frequencies = [18, 22, 20, 25, 17, 18]    # Face 1 to Face 6
total_rolls = sum(observed_frequencies)
expected_frequencies = [total_rolls / 6] * 6       # fair die: 20 per face
significance_level = 0.05

chi_square_statistic, p_value = chisquare(f_obs=observed_frequencies, f_exp=expected_frequencies)

print("Chi-Square Statistic: ", round(chi_square_statistic, 4))
print("P-Value:              ", round(p_value, 5))

if p_value < significance_level:
    print("We reject H0 (Null Hypo.) and consider Alternative Hypothesis")
else:
    print("We do not reject H0")
```

##### **Result:**
**Test:** Chi Square Goodness of Fit Test  
**H0:** Die is fair  
**H1:** Die is not fair    
**Business Conclusion:** At the 5% significance level, there is no evidence that the die is unfair. So the manufacture's claiim is accepted.
