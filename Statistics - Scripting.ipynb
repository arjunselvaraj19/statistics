## 1. A company claims that the average time taken to resolve customer tickets is 30 minutes. A random sample of 40 tickets shows a mean resolution time of 33 minutes. At the 5% level of significance, test whether the average resolution time has increased.

```python
#   H0: Current process/Statement of No Change - There is no significant difference in the average resolution time
#   H1: Statement of change/New process - There is a significant difference in the average resolution time
```

```python
from scipy import stats
import numpy as np
```

```python
#stats.ttest_1samp(a = [30,40,26,38,50], popmean,) # a - we need to provide the sample of all 40 it should come the avg 33

# Instead going like this alternative is below
```

```python
np.random.normal(loc=33, scale=5.0, size=40).astype("int")
#stats.ttest_1samp(a = [30,40,26,38,50], popmean,) 
```

```python
np.random.normal(loc=33, scale=5.0, size=40).astype("int").mean()
```

```python
observation_40_tickets = np.random.normal(loc=33, scale=5.0, size=40).astype("int")
observation_40_tickets
```

```python
t_stats, p_value = stats.ttest_1samp(a = observation_40_tickets, popmean = 30, alternative="greater")
t_stats, p_value
```

```python
if p_value < 0.05:
    print("We Reject H0")
else:
    print("We do not reject H0")
```

```python
# Simple way instead of above steps
np.random.normal(loc=33, scale=5.0, size=40).astype("int")
observation_40_tickets = np.random.normal(loc=33, scale=5.0, size=40).astype("int")
t_stats, p_value = stats.ttest_1samp(a = observation_40_tickets, popmean = 30, alternative="greater")
if p_value < 0.05:
    print("We Reject H0")
else:
    print("We do not reject H0")
```

### At 5% level of significance, we can reject the Null Hypotesis and claim that the average resolution time has increased.

### **Note**
1. P-Value
2. Level of Significance
3. How to break the a problem statement and understand in a better way and know how to choose which particular tests

## Imp:::  Alway P-Value by default Python program gives 2 Tailed Test Value

### So due to that, we need to divided by 2

```python
# Simple way instead of above steps
np.random.normal(loc=33, scale=5.0, size=40).astype("int")
observation_40_tickets = np.random.normal(loc=33, scale=5.0, size=40).astype("int")
t_stats, p_value = stats.ttest_1samp(a = observation_40_tickets, popmean = 30, alternative="greater")

p_value = p_value / 2 # this should apply for only if direction is right or left (in this example problem stmt says normal resolution is 30 mins but 40 ticktes takes 33 mins so it is increasing so it is right one tailed)

if p_value < 0.05:
    print("We Reject H0 (Null Hypo.)")
else:
    print("We do not reject H0")
```
